```python
class Human:
    # コンストラクタの定義
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# インスタンス化する
person = Human("侍太郎", 36)

# 値を出力する
person.printinfo()
```
