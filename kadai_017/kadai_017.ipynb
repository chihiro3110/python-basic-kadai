```python
ADULT_AGE = 20

class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):

        if self.age >= ADULT_AGE:
            print(f"{self.name}さんは大人です。")
        else:
            print(f"{self.name}さんは大人ではありません。")


def create_humans(data_list):
    humans = []


    for data in data_list:

        name = data["name"]
        age = data["age"]

        human = Human(name, age)
        humans.append(human)

    return humans


human_data = [
    {"name": "太郎", "age": 25},
    {"name": "花子", "age": 18},
    {"name": "次郎", "age": 30}
]

human_list = create_humans(human_data)

for person in human_list:
    person.check_adult()
```
